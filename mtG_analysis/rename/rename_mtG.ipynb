```python
import os
import csv
```

```python
os.listdir()
```

```python
def rename_fasta(file):
    print(file)
    with open(file, 'r', encoding = 'utf-8') as datafile:
        seqStr = datafile.read()
        seqList = seqStr.split('\n')
        seqHead = seqList[0]
        seqBody = ''.join(seqList[1:])
        datafile.close()
        
        neo_seqHead = seqHead + ' [location=mitochondrion]'
        
        
        with open(file, 'w', encoding = 'utf-8') as fastaFile:
            fastaFile.write(neo_seqHead)
            fastaFile.write('\n')
            fastaFile.write(seqBody)
            fastaFile.write('\n')
            fastaFile.close()
        print(neo_seqHead, len(seqBody))
```

```python
rename_fasta('AL101.fasta')
```

```python
file_count = 0

fastaList = os.listdir()

for file in fastaList:
    if file.endswith('.fasta')
    file_count += 1
    
print(file_count)
    
    
```

```python
fileExt = r'.fasta'
fastaList = [_ for _ in os.listdir() if _.endswith(fileExt)]

for file in fastaList:
    rename_fasta(file)
```
