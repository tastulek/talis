```python
# for reading a fasta file;

datafile = 'Jeti_CentEurAsia_mtG_N2154_bwa_aln_sorted_Jul20_21.fasta'

seqInfo_list = []

with open(datafile, 'r') as seqData:
    seqList = seqData.readlines()
    allSeq = ''.join(seqList)
    
    #print(len(allSeq))
    #print(allSeq[0:1000])
    indiSeq = allSeq.split('>')
    #for removing the first empty item created by split('>')
    indiSeq = indiSeq[1:]
    
    seq_dict = {}
    
    for seq in indiSeq:
        seq = seq.split('\n')
        seq_head = seq[0]
        seq_body = seq[1]
        seqSize = len(seq_body)
        
        seq_dict[seq_head] = seqSize
        #seq_dict['value'] = seqSize
        
        seqInfo_list.append(seq_dict)
           
    
    
    print(len(indiSeq))
    
```
