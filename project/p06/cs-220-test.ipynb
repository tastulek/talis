```python
import csv

# Modified from https://automatetheboringstuff.com/chapter14/
def process_csv(filename):
    example_file = open(filename, encoding="utf-8")
    example_reader = csv.reader(example_file)
    example_data = list(example_reader)
    example_file.close()
    return example_data

# Use process_csv to pull out the header and data rows
csv_rows = process_csv("airbnb.csv")
csv_header = csv_rows[0] # A list of the column headers
csv_data = csv_rows[1:] # The entire CSV data set besides the headers
```

```python
def cell(row_idx, col_name):
    col_idx = csv_header.index(col_name)
    val = csv_data[row_idx][col_idx]
    if val == "":
        return None
    # optional: convert types based on column name, ensuring that the float and int values are appropriately converted
    intList = ['room_id','host_id', 'price', 'minimum_nights', 'number_of_reviews', 'calculated_host_listings_count', 'availability_365']
    floatList = ['latitude','longitude', 'reviews_per_month']
    if col_name in intList:
        val = int(val)
        #print(type(val))
    elif col_name in floatList:
        #print(col_name)
        val = float(val)
        #print(type(val))
    
    return val
```

```python
#Q1: What neighborhood groups are present in the airbnb dataset?

all_items = []

def findElements(col_name):
    for i in range(len(csv_data[0:])):
        item = cell(i, col_name)
        all_items.append(item)
    # Removing duplicates by converting the list into set;
    all_items_set = set(all_items)
    
    # Convert the set back to list;
    all_items_final = list(all_items_set)
    
    return all_items_final

findElements('neighborhood_group')
```

```python
#Q2: What is the average number of reviews among all the reviews?

import math 

all_reviews = []
reviewSum = 0

def findMean(col_name):
    
    for i in range(len(csv_data[0:])):
        
        col_idx = csv_header.index(col_name)
        review = cell(i, col_name)
        
        # Collecting number of reviews as a list.
        all_reviews.append(review)
   
    #Adding all review numbers;
    for n in all_reviews:
        global reviewSum
        reviewSum += n
    #print('Sum:',reviewSum)
    
    #Finding reviews of how many apartments/houses being considered in the list;
    num_reviews = len(all_reviews) 
    #print('Nr. of apartments being reviewed:', num_reviews)
    
    #Finding the mean/average number of reviews for each apartment;
    mean_reviews = reviewSum / num_reviews
    
    #Removing the decimal places and returning the mean as an integer;
    mean_reviews = math.floor(mean_reviews)
    return mean_reviews
    


findMean('number_of_reviews')
```

```python
# This function is for answering the following questions;


def find_room_names(contained):
    rNameList = []

    # TODO: create a list
    for i in range(len(csv_data[0:])):
        col_name = 'name'
        rName = cell(i, col_name)
        
    # TODO: check if the room name contains the `contained` parameter (case insensitive)
        rName = str(rName)
        rName_lower = rName.lower()
        contained_lower = contained.lower()
        
        # TODO: add every unmodified room name to the return list
        if contained_lower in rName_lower:
            rNameList.append(rName)
            
    # TODO: return your list of room names
    return rNameList
```

```python
#Q6: Which host names are anagrams of the word "aisle"?

anagramList = []

def host_anagram(word):

    # TODO: create a list
    for i in range(len(csv_data[0:])):
        col_name = 'host_name'
        hostName = cell(i, col_name)
        hostName = str(hostName)
        
        anagram = ''
        
        if len(hostName) == len(word):
            
            for i in range(len(word)):
                hostName_lower = hostName.lower()
                word_lower = word.lower()
                
                if not (word_lower[i] in hostName_lower):
                    continue
                    
                elif word_lower[i] in hostName_lower:
                    if hostName_lower.count(word_lower[i]) == 1:
                        anagram += hostName[i]
        
        if anagram == hostName:
            anagramList.append(anagram)
            
    anagramSet = set(anagramList)
    anagramList_concise = list(anagramSet)
    return anagramList_concise


host_anagram('aisle')               
        
```

```python
#This function is for the following two questions;



def availability_per_host_name(host_name, neighborhood_group = None):
    
    vacantR_list = []
    vacantR_set = {}
    
        
    col_name_01 = 'host_name'
    col_name_02 = 'neighborhood_group'
    col_name_03 = 'availability_365'
    
    # TODO: create a list
    
    for i in range(len(csv_data[0:])):
        hostName = cell(i, col_name_01)
        nGroup = cell(i, col_name_02)
        vacancy = cell(i, col_name_03)
        vacancy = int(vacancy)
        
        
        # TODO: add every availability matching the host_name and neighborhood_group to the list as an int
        if hostName == host_name:
            if nGroup == neighborhood_group or neighborhood_group == None:
                if vacancy == 0:
                    vacantR_list.append(vacancy)
                 
                elif vacancy > 0:
                 
                    vacantR_list.append(vacancy)

    # For removing duplicates and sorting the list;
    vacantR_set = set(vacantR_list)
    finalR_list = list(vacantR_set)
    #sorted(finalR_list, reverse = True)
    
    # TODO: return your list of availabilities
    return sorted(finalR_list, reverse = True)
                   
```

```python
#This function is for the next several questions;



def find_prices_within(lat_min, lat_max, long_min, long_max):
    price_list = []
    

    col_name01 = 'latitude'
    col_name02 = 'longitude'
    col_name03 = 'price'

    # TODO: create a list
        
    for i in range(len(csv_data[0:])):
        lat_val = cell(i, col_name01)
        long_val = cell(i, col_name02)
        price_val = cell(i, col_name03)
        
        # TODO: add every price of rooms that locate in the given area to the list
        if lat_min <= lat_val <= lat_max:
            if long_min <= long_val <= long_max:
                price_list.append(price_val)
                
         
    # TODO: return the filled list of prices
    return price_list

#find_prices_within(40.50, 40.75, -74.00, -73.95)
```

```python
#Q15: What is the average ratio of the number of reviews to availability in the neighborhood Manhattan Beach?


def popularity(place):
    
    ratio_list = []
    
        
    col_name_01 = 'neighborhood'
    col_name_02 = 'number_of_reviews'
    col_name_03 = 'availability_365'
    
    # TODO: create a list
    
    for i in range(len(csv_data[0:])):
        nH = cell(i, col_name_01)
        if nH == place:
            
            nReview = cell(i, col_name_02)
            vacancy = cell(i, col_name_03)
            
            if nReview == None:
                continue
                
            elif vacancy == 0 or vacancy == None:
                continue
                
            else:
                
                vacancy = int(vacancy)
                nReview = int(nReview)

                ratio = nReview / vacancy

                ratio_list.append(ratio)
    
    ratioSum = 0  
    
    for i in range(len(ratio_list)):
        ratioSum += ratio_list[i]
        
    meanRatio = ratioSum / len(ratio_list)
    
    return meanRatio

popularity('Manhattan Beach')
```

```python
#This function is for the next two questions;

def secondary_word_in_found_rooms(name_w01, name_w02):
    
    intersection_list = []
    name_w01_list = find_room_names(name_w01)
    #print(name_w01_list[0:10])
    
    name_w02_list = find_room_names(name_w02)
    #print(name_w02_list[0:10])
    
    for name in name_w01_list:
        if name in name_w02_list:
            intersection_list.append(name)
    
    #print(intersection_list)
    numerator = len(intersection_list)  
    denominator = len(name_w01_list)
    
    wordRatio = numerator / denominator * 100
    
    return wordRatio
        
    

```

```python
#Q20: What is the minimum amount of money one needs to spend to stay for 10 days 
# in Manhattan, and then 5 days in Staten Island?



def costEval(nG01, nG02, dur01 = 10, dur02 = 5):

    col_name_01 = 'neighborhood_group'
    col_name_02 = 'availability_365'
    col_name_03 = 'minimum_nights'
    col_name_04 = 'price'
    
    
    priceList01 = []
    priceList02 = []
    
    for i in range(len(csv_data[0:])):
        nGroup = cell(i, col_name_01)
        vacancy = cell(i, col_name_02)
        min_nights = cell(i, col_name_03)
        room_price = cell(i, col_name_04)
        
        if nGroup == nG01:
            if vacancy == 0 or vacancy == None:
                continue
                
            elif min_nights == 0 or min_nights == None:
                continue
            
            elif vacancy >= dur01 and min_nights <= dur01:
                
                if room_price == 0 or room_price == None:
                    
                    continue
                    
                else:
                    
                    priceList01.append(room_price)
                    
        elif nGroup == nG02:
            if vacancy == 0 or vacancy == None:
                continue
                
            elif min_nights == 0 or min_nights == None:
                continue
            
            elif vacancy >= dur02 and min_nights <= dur02:
                
                if room_price == 0 or room_price == None:
                    continue
                
                else:
                    
                    priceList02.append(room_price)
    
    #print(priceList01)
    priceList01 = sorted(priceList01)
    lowestPrice01 = priceList01[0]
    
    
    #print(priceList02)
    priceList02 = sorted(priceList02)
    lowestPrice02 = priceList02[0]
    
    totalCost = lowestPrice01 * dur01 + lowestPrice02 * dur02
    
    return totalCost
        
                    
        
        
                

costEval('Manhattan', 'Staten Island')                   
                
                
                

    
```
