```python
#Function 1: year_max(year)
# This function is for finding maximum amount of energy consumed
# in a given year when all engery sources being considered;


def year_max(year):
    # grab the consumption of each energy source in the given year
    hydro_use = project.get_consumption(hydroIndex, year)
    solar_use = project.get_consumption(solarIndex, year)
    wind_use = project.get_consumption(windIndex, year)
    wood_use = project.get_consumption(woodIndex, year)
    bio_use = project.get_consumption(bioIndex, year)
    
    return max(hydro_use, solar_use, wind_use, wood_use, bio_use)
```

```python
# Function 2: energy_min(source)
# This function is for finding minimum amount of energy for a given 
# energy source in all years;

def energy_min (source):
    #getting the index for the source of energy:
    source_index = project.get_idx(source)
    
    #getting the amount of energy used each year:
    y15 = project.get_consumption(source_index, 2015)
    y16 = project.get_consumption(source_index, 2016)
    y17 = project.get_consumption(source_index, 2017)
    y18 = project.get_consumption(source_index, 2018)
    y19 = project.get_consumption(source_index, 2019)
    y20 = project.get_consumption(source_index, 2020)
    
    return min(y15, y16, y16, y17, y18, y19, y20)
```

```python
#Function 3: energy_avg(energy_source)
#This function should compute the average energy consumed by the US electric power sector from 
#the given energy source across the six years in the datatset.

def energy_avg(energy_source):
    #getting the index for the source of energy:
    source_index = project.get_idx(energy_source)
    
    #getting the amount of energy used each year:
    y15 = project.get_consumption(source_index, 2015)
    y16 = project.get_consumption(source_index, 2016)
    y17 = project.get_consumption(source_index, 2017)
    y18 = project.get_consumption(source_index, 2018)
    y19 = project.get_consumption(source_index, 2019)
    y20 = project.get_consumption(source_index, 2020)
    
    average = (y15 + y16 + y17 + y18 + y19 + y20)/6
    
    return average
```

```python
# Function 4: year_sum(year)
# This function should compute the total renewable energy consumption of the US electric power sector 
#(over the five sources) in a given year.

def year_sum(year = 2016):
    
    hydro_use = project.get_consumption(hydroIndex, year)
    solar_use = project.get_consumption(solarIndex, year)
    wind_use = project.get_consumption(windIndex, year)
    wood_use = project.get_consumption(woodIndex, year)
    bio_use = project.get_consumption(bioIndex, year)
    
    all_use = (hydro_use + solar_use + wind_use + wood_use + bio_use)
    
    return all_use 
```

```python
#Function 5: change_per_year(energy, start_year, end_year)
# This function should return the average increase in consumption (could be negative if there's a decrease) over the period from start_year to end_year for 
# the given energy source (example energy source: "Solar Energy").

def change_per_year(energy, start_year = 2015, end_year = 2020):
    
    #getting the index for the source of energy:
    source_index = project.get_idx(energy)
    
    ySpan = end_year - start_year
    
    #getting the amount of energy used in the start year and end year:
    yStart = project.get_consumption(source_index, start_year)
    yEnd = project.get_consumption(source_index, end_year)
    
    average_change = (yEnd - yStart) / ySpan
    
    return average_change
       
```

```python
#Q13: 
# How much has the consumption of Solar Energy by the US electric power sector increased per year (on average) from 2015 to 2020?
change_per_year('Solar Energy')
```

```python
#Q14: 
#How much has the consumption of Wind Energy by the US electric power sector 
#increased per year (on average) from 2016 to 2020?
change_per_year('Wind Energy', 2016)
```

```python
#Function 6: find_threshold_year(energy, threshold)
import math

def find_threshold_year(energy, threshold=0):
    
    # TODO: compute the average change in consumption from 2015 to 2020
    av_change = change_per_year(energy)
    #print(av_change)
    # TODO: assume that the energy consumption keeps decreasing at same rate
    
    # TODO: compute the required decrease in energy consumption from 2020 to cross threshold
    e_use_20 = project.get_consumption(project.get_idx(energy), 2020)
    #print(e_use_20)
    
    # TODO: estimate the number of years it will take from 2020 to cross threshold
    
    #e_decrease = threshold - e_use_20
    #print(e_decrease)
    
    #e_decrease = abs(e_decrease)
    
    year_needed = (threshold - e_use_20) / av_change
    
    #print(year_needed)
    
    year_estimate = 2020 + math.ceil(year_needed)
    
    
    return year_estimate
    
    
    # TODO: use math.ceil() to round it up.
```

```python
#Function 7: find_overtake_year(energy1,energy2)


def find_overtake_year(energy1, energy2):

    # TODO: compute the average change in consumption for both energy sources from 2015 to 2020.
    av_change_1 = change_per_year(energy1)
    av_change_2 = change_per_year(energy2)
    
    e_change_gap = av_change_2 - av_change_1
    
    #print(av_change_1)
    #print(av_change_2)
    # TODO: assume both energy consumptions keep changing at the same rate.
    
    # TODO: compute the intial gap between consumption of energy1 and energy2 in 2020. 
    e1_use_20 = project.get_consumption(project.get_idx(energy1), 2020)
    e2_use_20 = project.get_consumption(project.get_idx(energy2), 2020)
    
    e_use_gap20 = e1_use_20 - e2_use_20
    
    year_needed = e_use_gap20 / e_change_gap
    
    
    # TODO: estimate the number of years it will take from 2020 for this gap to decrease to 0.
    
    # TODO: use math.ceil() to round it up.
    year_estimate = 2020 + math.ceil(year_needed)
    return year_estimate
find_overtake_year('Biomass Energy', 'Wood Energy')
```
