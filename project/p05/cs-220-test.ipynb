```python
#Q19: Which of two hurricanes took longer to dissipate: Dora or Karen?

import datetime

def get_number_of_days(start_date, end_date):
    """Gets the number of days between the start_date (in 'mm/dd/yyyy' format) and end_date (in 'mm/dd/yyyy' format)"""
    day1 = datetime.datetime.strptime(start_date, '%m/%d/%Y') # The second argument is a format string to tell the function how to process the date string
    day2 = datetime.datetime.strptime(end_date, '%m/%d/%Y')
    delta = day2 - day1
    return delta.days

def compare(hurricane_01,hurricane_02):
    
    for i in range(project.count()):
        if project.get_name(i) == hurricane_01:
            index_01 = i
            date_formed_01 = project.get_formed(index_01)
            date_dissipated_01 = project.get_dissipated(index_01)
            
            days_01 = get_number_of_days(date_formed_01, date_dissipated_01)
            
        elif project.get_name(i) == hurricane_02:
            index_02 = i
            date_formed_02 = project.get_formed(index_02)
            date_dissipated_02 = project.get_dissipated(index_02)
            
            days_02 = get_number_of_days(date_formed_02, date_dissipated_02)
    if days_01 > days_02:
        return hurricane_01
    elif days_01 < days_02:
        return hurricane_02
    else:
        'Maybe Draw!'

compare('Dora', 'Karen')
```
