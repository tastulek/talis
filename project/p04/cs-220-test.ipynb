```python
# Damage function;
import project

def damage(attacker, defender):
    physical_damage = 10 * project.get_attack(attacker) / project.get_defense(defender)
    special_damage = 10 * project.get_sp_atk(attacker) / project.get_sp_def(defender)
    if physical_damage > special_damage:
        return physical_damage
    else:
        return special_damage
```

```python
# Here is a new function,type_bonus, which calculates the type bonus of an attack_type against a defender.
bonus = ''

def type_bonus(attack_type, defender):
    defender_type1 = project.get_type1(defender)
    defender_type2 = project.get_type2(defender)
    bonus = project.get_type_effectiveness(attack_type, defender_type1)
    
    if defender_type2 != 'None':
        
        bonus2 = project.get_type_effectiveness(attack_type, defender_type2)
        bonus = bonus * bonus2
        
    return bonus
```

```python
# This function caclulates Effective Damage, i.e. compute the actual damage that an attacker would do to the defender, 
# taking into account, both the attack mode and attack type.

def effective_damage(attacker, defender):

    #TODO: check if attacker has two types
    def get_num_types(attacker):
    
        if project.get_type1(attacker) == 'None':
            return 0
        elif project.get_type2(attacker) == 'None':
            return 1
        else:
            return 2
    
    #TODO: compute the bonus of attacker's type(s) against defender
    if get_num_types(attacker) == 1:
        bonus = type_bonus(project.get_type1(attacker), defender)
        
    elif get_num_types(attacker) == 2:
        bonus1 = type_bonus(project.get_type1(attacker), defender)
        bonus2 = type_bonus(project.get_type2(attacker), defender)
        
        
    #TODO: find the attack_type with the higher bonus
        bonus = max(bonus1, bonus2)
        
    #TODO: compute the damage caused by attack, considering the higher bonus and return it
    effectiveDamage = damage(attacker, defender) * bonus
    
    return effectiveDamage
```

```python
# This function is for calculating how many hits each Pokémon can take before fainting
import math

def num_hits(attacker, defender):
    maxHit = project.get_hp(defender)/ effective_damage(attacker, defender)
    maxHit = math.ceil(maxHit)
    return maxHit
```

```python
# This function is for simulating battle!

def battle(pkmn1, pkmn2):

    #TODO: Return the name of the pkmn that can take more hits from the other
    # pkmn. If both pkmn faint within the same number of moves, return the
    # string 'Draw'
    pkmn1_defend_maxHit = num_hits(pkmn2, pkmn1)
    pkmn2_defend_maxHit = num_hits(pkmn1, pkmn2)
    
    #For calculating the difference between the two pokemon
    # in terms of maximum number of hits;
    
    hitDif = pkmn1_defend_maxHit - pkmn2_defend_maxHit
    
    if hitDif > 10:
        return pkmn2 + ' ran away'
    
    elif hitDif < -10:
        return pkmn1 + ' ran away'

    #This block is for comparing which pokemon can receive the most number of hits;
    if pkmn1_defend_maxHit > pkmn2_defend_maxHit:
        return pkmn1
    
    elif pkmn1_defend_maxHit < pkmn2_defend_maxHit:
        return pkmn2
    
    # This block takes speed into consideration;
    
    elif pkmn1_defend_maxHit == pkmn2_defend_maxHit:
        speed01 = project.get_speed(pkmn1)
        speed02 = project.get_speed(pkmn2)
        
        if speed01 > speed02:
            return pkmn1
        
        elif speed01 < speed02:
            return pkmn2
        
        else: 
            return 'Draw'
```

```python
# This the final function of P04, that is for team battle;

def two_on_two_battle(pkmn1, pkmn2, pkmn3, pkmn4):
    
    # TODO: implement the first round
    winner01 = battle(pkmn1, pkmn3)
    winner02 = battle(pkmn2, pkmn4)
    winner03 = battle(pkmn1, pkmn4)
    winner04 = battle(pkmn2, pkmn3)
    
    # For finding the winner in the 1st team battle in the first round;
    if winner01 == 'Draw':
        return 'No Results'
    elif winner01 == pkmn1 + ' ran away':
        return 'No Results'
    elif winner01 == pkmn2 + ' ran away':
        return 'No Results'
    #else:
        #print(winner01)
    
    # For finding the winner in the 2nd team battle in the first round;
    
    if winner02 == 'Draw':
        return 'No Results'
    elif winner02 == pkmn1 + ' ran away':
        return 'No Results'
    elif winner02 == pkmn2 + ' ran away':
        return 'No Results'
    #else:
        #print(winner02)
        
    # For determinging which team is the winner in the 1st round;
    
    if winner01 == pkmn1 and winner02 == pkmn2:
        return 'Team A wins'
    
    elif winner01 == pkmn3 and winner02 == pkmn4:
        return 'Team B wins'
    
    
    # TODO: implement the second round
    if winner01 == pkmn1 and winner02 == pkmn4:
        winner03 = battle(pkmn1, pkmn4)

    elif winner01 == pkmn3 and winner02 == pkmn2:
        winner04 = battle(pkmn3, pkmn2)
        
    # For determinging which team is the winner in the 2nd round;
    
    if winner03 == pkmn1 or winner04 == pkmn2:
        return 'Team A wins'
    
    elif winner03 == pkmn4 or winner04 == pkmn3:
        return 'Team B wins'

```
