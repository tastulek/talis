```python

# Modified from https://automatetheboringstuff.com/chapter14/
def process_csv(filename):
    example_file = open(filename, encoding="utf-8")
    example_reader = csv.reader(example_file)
    example_data = list(example_reader)
    example_file.close()
    return example_data

allData = process_csv('vaccinations.csv')
data_header = allData[0]
data_body = allData[1:]
```

```python
def findCell(row_index, col):
    
    col_index = data_header.index(col)
    val = data_body[row_index][col_index]
    
    intList = ['daily_vaccinations','total_vaccinations','people_vaccinated','people_fully_vaccinated', 'population']
    if val == '':
        return None
        
    if col in intList:
        val = int(val)
        
    return val
    
          
```

```python
#Q1: What dates are mentioned in the dataset?


def dateRange():
    date_list = []
    for i in range(len(data_body)):

        Date = findCell(i, 'date')
        date_list.append(Date)

    date_set = set(date_list)
    date_shortList = list(date_set)
    date_shortList.sort()

    return date_shortList
        
dateRange()
```

```python
#Q2: What countries are listed in the dataset?

country_list = []

def findCountry():
    
    for i in range(len(data_body)):
        
        nation = findCell(i, 'country')

        country_list.append(nation)

    country_set = set(country_list)
    country_shortList = list(country_set)
    country_shortList.sort()
    nation_count = len(country_shortList)
    print('There are', nation_count, 'countries in the datafile.')
    
    return country_shortList
        
findCountry()
```

```python
#Q3: What are the populations of all the countries in the dataset?

popSize = {} #key: country name, value: popluation

def find_popSize():
    for i in range(len(data_body)):
        country = findCell(i, 'country')
        population = findCell(i, 'population')
        
        if country not in popSize:
            popSize[country] = population # Initial place holder for country, we will replace this using next condition
            # if popSize[country] = None is used, it causes symantic errors;
            
        if population != None: # TODO: check that the population data is not missing
            popSize[country] = population # TODO: add the key-value pair to the dict
            

    return popSize

find_popSize()
```

```python
#Q4: How many countries have daily vaccination date for the date '10/12/2021'?

#Create an empty list to track country names
nationList = []
#Iterate over the data
def find_country_count(vac_date):
    
    for i in range(len(data_body)):
        
        #use cell function calls to retrieve relevant column information
        vacDate = findCell(i, 'date')
        country = findCell(i, 'country')
        vacData = findCell(i, 'daily_vaccinations')
        
        #check for date match
        #if vaccination data is not missing, add country to the list
        if vacDate == vac_date:
            if vacData == None:
                continue
                
            else:
                nationList.append(country)
    
    #Get unique country names - what data structure will be useful here?
    #How can you find the count of unique country names?
    #print('rows:', len(data_body))
    #print('NationList:', len(nationList))
    #print(nationList)
    nationSet = set(nationList) 
    final_NationList = list(nationSet)
    country_count = len(final_NationList)
    
    return country_count


find_country_count('10/12/2021')
```

```python
# This function is for the following several questins;

def daily_vaccinations_on(date):
    '''return a dictionary mapping each country to the number of daily vaccinations on the given date; 
    if data is missing, the value for that country is None'''
    
    daily_vax_dict = {}
    
    for i in range(len(data_body)):
        country_i = findCell(i, 'country')
        date_i = findCell(i, 'date')
        daily_vax_i = findCell(i, 'daily_vaccinations')
        
        if date_i == date: # TODO: check if date_i is the correct date 
            # TODO: map the country to the number of daily vaccinations on that day
            daily_vax_dict[country_i] = daily_vax_i
            
    return daily_vax_dict
```

```python
#Q6: How many vaccinations are estimated to have been done for each country 
# across the dates mentioned in the dataset?

def vacCount():
    
    vac_grandList = []
    vac_grandDict = {}
    
    # find date ranges of the week;
    vac_dateRange = dateRange()
    
    for vac_date in vac_dateRange:
        vac_dict = daily_vaccinations_on(vac_date)
        vac_grandList.append(vac_dict)
        
    for indi_dict in vac_grandList:
        
        
        
        for key, val in indi_dict.items():

            
            if key not in vac_grandDict:
                vac_grandDict[key] = val
                
            else:
  
                # to make sure that the value in the vac_grandDict 
                # and the val in indi_dic are not empty, i.e. None;
                if vac_grandDict[key] != None and val != None: 
                    
                    vac_grandDict[key] += val 
                    
                #If there is no record on the first recorded date, then replace it with new val;
                if vac_grandDict[key] == None and val != None:
                    
                    vac_grandDict[key] = val
                    
                     
        
    return vac_grandDict
        
vacCount()
```

```python
#Q7: Which country is estimated to have completed the most number of vaccinations 
# across the dates mentioned in the dataset?

def mostVac():
    
    vac_Data = vacCount()
    
    highest_vac = 0
    bestVac_nation = ''
    
    for country in vac_Data:
        
        # To skip the countries without any data;
        if vac_Data[country] == None:
            continue
            
        # To find the country with highest vaccinations;   
        if vac_Data[country] > highest_vac:
            highest_vac = vac_Data[country]
            
            bestVac_nation = country
            
    return bestVac_nation

mostVac()   
```

```python
#Q8: How many vaccinations are estimated to have been done across the whole world 
# on each date mentioned in the dataset?

def vacStats():
    
    vac_grandList = []
    date_dict = {}
    
    # find date ranges of the week;
    vac_dateRange = dateRange()
    
    for vac_date in vac_dateRange:
        vac_dict = daily_vaccinations_on(vac_date)
        
        # For removing the countries without any data available on a particular date;
        updated_vac_dict = {key: value for key, value in vac_dict.items() if value is not None}
        vac_dict.clear()
        vac_dict.update(updated_vac_dict)
        
        date_dict[vac_date] = sum(vac_dict.values())
    vac_grandList.append(date_dict)
    
    for item in vac_grandList:
        vac_grandDict = item
        
    return vac_grandDict


vacStats()
```

```python
#Q9: For each country in the dataset, when is the last day 
# for which we have data about the number of fully vaccinated people?

def find_final_vacDate():
    
    max_vac = 0
    final_vac_date = ''
    final_date_dict = {}
    
    for i in range(len(data_body)):
        fully_vac_i = findCell(i, 'people_fully_vaccinated')
        date_i = findCell(i, 'date')
        country_i = findCell(i, 'country')
        
        # for including countries for which no data is available;
        if country_i not in final_date_dict:
            final_date_dict[country_i] = None
            
        if fully_vac_i == None:
            continue
            
        if fully_vac_i > max_vac:
            max_vac = max_vac
            final_vac_date = date_i
            final_date_dict[country_i] = final_vac_date
            
    return final_date_dict
        
    
find_final_vacDate()
```

```python
import datetime

def get_number_of_days(start_date, end_date):
    """Gets the number of days between the start_date (in 'mm/dd/yyyy' format) and end_date (in 'mm/dd/yyyy' format)"""
    day1 = datetime.datetime.strptime(start_date, '%m/%d/%Y') # The second argument is a format string to tell the function how to process the date string
    day2 = datetime.datetime.strptime(end_date, '%m/%d/%Y')
    delta = day2 - day1
    
    return delta.days
```

```python
#This function is for following several questions;

def fully_vaccinated_by(date):
    '''return a dictionary mapping each country to the most recent number of people 
    fully vaccinated by the given date; if no data is available, the value is None.'''
    
    fully_vax_dict = {}
    
    for i in range(len(data_body)):
        
        country_i = findCell(i, 'country')
        date_i = findCell(i, 'date')
        fully_vax_i = findCell(i, 'people_fully_vaccinated')
        
        if country_i not in fully_vax_dict:
            fully_vax_dict[country_i] = None
            
        if get_number_of_days(date_i, date) >= 0:
            if fully_vax_i != None:
                fully_vax_dict[country_i] = fully_vax_i
    
    return fully_vax_dict

```

```python
#This function is for getting total vaccination number on a given date;

def total_vaccinations_by(date):
    
    total_vax_dict = {}
    
    for i in range(len(data_body)):
        
        country_i = findCell(i, 'country')
        date_i = findCell(i, 'date')
        total_vax_i = findCell(i, 'total_vaccinations')
        
        if country_i not in total_vax_dict:
            total_vax_dict[country_i] = None
            
        if get_number_of_days(date_i, date) >= 0:
            if total_vax_i != None:
                total_vax_dict[country_i] = total_vax_i
    
    return total_vax_dict



#total_vaccinations_by('10/14/2021')
```

```python
#This function is for getting the number of people vaccinated on a given date;

def people_vaccinated_by(date):
    
    people_vax_dict = {}
    
    for i in range(len(data_body)):
        
        country_i = findCell(i, 'country')
        date_i = findCell(i, 'date')
        people_vax_i = findCell(i, 'people_vaccinated')
        
        if country_i not in people_vax_dict:
            people_vax_dict[country_i] = None
            
        if get_number_of_days(date_i, date) >= 0:
            if people_vax_i != None:
                people_vax_dict[country_i] = people_vax_i
    
    return people_vax_dict



#people_vaccinated_by('10/14/2021')
```

```python
# This function is for the following questions;



def create_dataStructure():
    vaccination_stats = {}

    populations = find_popSize() # a dict mapping each country to population of country
    date_range = dateRange()
    nationList = findCountry()
    
    for date in date_range:
        daily_vax_dict = daily_vaccinations_on(date) # a dict mapping each country to number of daily_vaccinations on date
        total_vax_dict = total_vaccinations_by(date) # a dict mapping each country to (interpolated) number of total_vaccinations on date
        people_vax_dict = people_vaccinated_by(date) # a dict mapping each country to (interpolated) number of people_vaccinated on date
        fully_vax_dict = fully_vaccinated_by(date) # a dict mapping each country to (interpolated) number of people_fully_vaccinated on date

        for country in nationList:
            if country not in vaccination_stats:
                vaccination_stats[country] = {}
            vaccination_stats[country][date] = {}
            vaccination_stats[country][date]['country'] = country
            vaccination_stats[country][date]['date'] = date
            vaccination_stats[country][date]['daily_vaccinations'] = daily_vax_dict[country]
            vaccination_stats[country][date]['total_vaccinations'] = total_vax_dict[country]
            vaccination_stats[country][date]['people_vaccinated'] = people_vax_dict[country]
            vaccination_stats[country][date]['people_fully_vaccinated'] = fully_vax_dict[country]
            vaccination_stats[country][date]['population'] = populations[country]

    return vaccination_stats

#create_dataStructure()
```

```python
#Q17: What is the estimated ratio of people who have been partly vaccinated 
# to people who have been fully vaccinated in each country, as of '10/16/2021'?

def vacRatio(date):
    
    vacRatio_dict = {}
    
    vacStats = create_dataStructure()
    
    for country in vacStats:
    
        partly_vac = vacStats[country][date]['people_vaccinated']

        fully_vac = vacStats[country][date]['people_fully_vaccinated']
        
        if partly_vac == None or fully_vac == None:
            vacRatio_dict[country] = None
        else:
            vacRatio_dict[country] = partly_vac / fully_vac - 1
            
    return vacRatio_dict
            
            
            
vacRatio('10/16/2021')            
```

```python
#Q19: Which country is estimated to have administered the most vaccines per capita, as of '10/14/2021'?

def per_capita_vac(date):
    
    
    per_capita_vac_dict = {}
    
    vacStats = create_dataStructure()
    
    mostVac = 0
    
    for country in vacStats:
    
        total_vac = vacStats[country][date]['total_vaccinations']

        pop = vacStats[country][date]['population']
        
        if total_vac == None or pop == None:
            continue
            
        else:
            per_capita_vac_dict[country] = total_vac / pop
            
    for country in per_capita_vac_dict:
        perCapita_vac = per_capita_vac_dict[country]
        
        if perCapita_vac > mostVac:
            mostVac = perCapita_vac
        
        if perCapita_vac == mostVac:
            
            nation_highestVac = country 
        
    return nation_highestVac


per_capita_vac('10/14/2021')    
```

```python
#Q20: Which country is estimated to have vaccinated the most people 
# as a percentage of its population on the date '10/10/2021'?

def vacRate(date):
    
    
    vacRate_dict = {}
    
    vacStats = create_dataStructure()
    
    highestPercent = 0
    
    for country in vacStats:
    
        daily_vac = vacStats[country][date]['daily_vaccinations']

        pop = vacStats[country][date]['population']
        
        if daily_vac == None or pop == None:
            continue
            
        else:
            vacRate_dict[country] = daily_vac / pop * 100
            
    for country in vacRate_dict:
        vac_rate = vacRate_dict[country]
        
        if vac_rate > highestPercent:
            highestPercent = vac_rate
        
        if vac_rate == highestPercent:
            
            nation_highest_vacRate = country 
        
    return nation_highest_vacRate


vacRate('10/10/2021')    
```
